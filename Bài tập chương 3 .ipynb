```python
#TH3.2
diem_goc = [7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]
diem_sap_xep = sorted(diem_goc)
n = 2
diem_cat = diem_sap_xep[n : len(diem_sap_xep) - n]
tb_cat = sum(diem_cat) / len(diem_cat) if diem_cat else 0
tb_thuong = sum(diem_goc) / len(diem_goc)

print("Trung bình cắt đầu - cuối ")
print(f"Danh sách gốc (được giữ nguyên): {diem_goc}")
print(f"Danh sách sau khi cắt bỏ 2 đầu: {diem_cat}")
print(f"Trung bình cắt đầu-cuối: {tb_cat:.2f}")
print(f"Trung bình thông thường: {tb_thuong:.2f}")
```

```python
#TH3.4
import statistics

du_lieu = [(1, 4.0), (2, 5.5), (3, 6.0), (4, 7.5), (5, 8.0)]
x = [item[0] for item in du_lieu]
y = [item[1] for item in du_lieu]
x_bar = sum(x) / len(x)
y_bar = sum(y) / len(y)
tu_so = sum((x[i] - x_bar) * (y[i] - y_bar) for i in range(len(x)))
mau_so = sum((x[i] - x_bar) ** 2 for i in range(len(x)))

b = tu_so / mau_so
a = y_bar - b * x_bar
gio_on = 6
diem_du_doan = a + b * gio_on

print("Hồi quy tuyến tính ")
print(f"Phương trình hồi quy thủ công: y = {a:.4f} + {b:.4f} * x")
print(f"Dự đoán điểm cho sinh viên ôn {gio_on} giờ: {diem_du_doan:.2f}")
b_lib, a_lib = statistics.linear_regression(x, y)
print(f"Kiểm tra bằng statistics.linear_regression: y = {a_lib:.4f} + {b_lib:.4f} * x")
```

```python
#TH3.6
from collections import defaultdict

danh_sach_sv = [
    ("An", 9.0), ("Bình", 7.5), ("Chi", 8.6), 
    ("Dung", 5.0), ("Em", 6.0), ("Phúc", 3.5), ("Giang", 7.0)
]

def quy_doi_diem(diem):
    if diem >= 8.5: return 'A'
    elif diem >= 7.0: return 'B'
    elif diem >= 5.5: return 'C'
    elif diem >= 4.0: return 'D'
    else: return 'F'
nhom_theo_chu = defaultdict(list)

for ten, diem in danh_sach_sv:
    diem_chu = quy_doi_diem(diem)
    nhom_theo_chu[diem_chu].append(ten)

print(" Gom nhóm học lực sinh viên ")
for key in sorted(nhom_theo_chu.keys()):
    print(f"Điểm {key}: {nhom_theo_chu[key]}")
```

```python
#TH3.8
cua_hang_1 = {"ao": 120, "quan": 80, "non": 30}
cua_hang_2 = {"ao": 90, "giay": 150, "non": 20}

tong_doanh_thu = cua_hang_1.copy()
for sp, doanh_thu in cua_hang_2.items():
    if sp in tong_doanh_thu:
        tong_doanh_thu[sp] += doanh_thu
    else:
        tong_doanh_thu[sp] = doanh_thu
doanh_thu_giam_dan = sorted(tong_doanh_thu.items(), key=lambda item: item[1], reverse=True)
khoa_1 = set(cua_hang_1.keys())
khoa_2 = set(cua_hang_2.keys())

ban_ca_hai = khoa_1.intersection(khoa_2)        
chi_ban_mot_cua_hang = khoa_1.symmetric_difference(khoa_2) 
print("--- TH 3.8: Phân tích doanh thu cửa hàng ---")
print("Tổng doanh thu theo sản phẩm (giảm dần):", doanh_thu_giam_dan)
print("Sản phẩm bán ở cả hai cửa hàng:", ban_ca_hai)
print("Sản phẩm chỉ bán ở một cửa hàng:", chi_ban_mot_cua_hang)
```

```python
#TH3.10
N = 1000
ung_vien = set(range(2, N + 1))
current = 2
while current * current <= N:
    if current in ung_vien:
        boi = current * 2
        while boi <= N:
            if boi in ung_vien:
                ung_vien.remove(boi)
            boi += current
    current += 1

so_nguyen_to = ung_vien

so_luong = len(so_nguyen_to)
tong_gia_tri = sum(so_nguyen_to)
cap_sinh_doi = []
for p in sorted(so_nguyen_to):
    if p + 2 <= N and (p + 2) in so_nguyen_to:
        cap_sinh_doi.append((p, p + 2))

print("Sàng Eratosthenes tìm số nguyên tố ")
print(f"Số lượng số nguyên tố <= 1000: {so_luong} (Khớp kết quả chuẩn: {so_luong == 168})")
print(f"Tổng giá trị các số nguyên tố <= 1000: {tong_gia_tri} (Khớp kết quả chuẩn: {tong_gia_tri == 76127})")
print(f"Số lượng cặp số nguyên tố sinh đôi: {len(cap_sinh_doi)}")
print(f"Danh sách các cặp sinh đôi: {cap_sinh_doi}")
```
