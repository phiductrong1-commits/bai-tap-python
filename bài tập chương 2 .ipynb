```python
print("=== CHƯƠNG TRÌNH TÍNH LƯƠNG TUẦN ===")

while True:
    nhap_vao = input("Nhập số giờ làm việc trong tuần: ")
    hop_le = True
    if nhap_vao == "":
        hop_le = False
    else:
        so_dau_tru = 0
        so_dau_cham = 0
        for i in range(len(nhap_vao)):
            ky_tu = nhap_vao[i]
            if ky_tu == '-':
                if i != 0: 
                    hop_le = False
                else:
                    so_dau_tru += 1
            elif ky_tu == '.':
                so_dau_cham += 1
            elif ky_tu < '0' or ky_tu > '9': 
                hop_le = False
        if so_dau_tru > 1 or so_dau_cham > 1:
            hop_le = False
    if hop_le == True:
        gio_lam = float(nhap_vao)
        if gio_lam < 0:
            print("Lỗi: Không được nhập số âm! Vui lòng nhập lại.")
        else:
            break
    else:
        print("Lỗi: Giá trị vừa nhập không phải là số hợp lệ! Vui lòng nhập lại.")
muc_luong_1h = 100000
gio_chuan = 40
he_so_vuot = 1.5

if gio_lam <= gio_chuan:
    tong_luong = gio_lam * muc_luong_1h
else:
    gio_vuot = gio_lam - gio_chuan
    tong_luong = (gio_chuan * muc_luong_1h) + (gio_vuot * muc_luong_1h * he_so_vuot)

print("Số giờ làm việc:", gio_lam)
print("Tổng lương tuần:", tong_luong, "VNĐ")
```

```python
print("=== CHƯƠNG TRÌNH THỐNG KÊ SỐ THỰC ===")
print("Nhập các số thực (gõ 'xong' để kết thúc):")

dem = 0
tong = 0.0
gia_tri_max = 0.0
gia_tri_min = 0.0

while True:
    nhap = input("Nhập số: ")
    if nhap == "xong" or nhap == "XONG" or nhap == "Xong":
        break
    hop_le = True
    if nhap == "":
        hop_le = False
    else:
        so_dau_cham = 0
        so_dau_tru = 0
        for i in range(len(nhap)):
            kt = nhap[i]
            if kt == '-':
                if i != 0:
                    hop_le = False
                else:
                    so_dau_tru += 1
            elif kt == '.':
                so_dau_cham += 1
            elif kt < '0' or kt > '9':
                hop_le = False
        if so_dau_cham > 1 or so_dau_tru > 1:
            hop_le = False
    if hop_le == False:
        print("Bỏ qua giá trị không phải số:", nhap)
        continue
    so = float(nhap)
    if dem == 0:
        tong = so
        gia_tri_max = so
        gia_tri_min = so
        dem = 1
    else:
        tong = tong + so
        dem = dem + 1
        if so > gia_tri_max:
            gia_tri_max = so
        if so < gia_tri_min:
            gia_tri_min = so
print("\n--- KẾT QUẢ THỐNG KÊ ---")
if dem == 0:
    print("Chưa có dữ liệu.")
else:
    trung_binh = tong / dem
    print("Số lượng giá trị hợp lệ:", dem)
    print("Tổng các số:", tong)
    print("Trung bình cộng:", trung_binh)
    print("Giá trị lớn nhất:", gia_tri_max)
    print("Giá trị nhỏ nhất:", gia_tri_min)
```

```python
print("=== DÃY COLLATZ ===")
while True:
    try:
        n = int(input("Nhập số nguyên dương n (ví dụ 29): "))
        if n > 0:
            break
        print("Lỗi: Phải nhập số lớn hơn 0!")
    except ValueError:
        print("Lỗi: Không phải số nguyên hợp lệ!")

n_ban_dau = n
so_buoc = 0
gia_tri_max = n
while n > 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1
        
    so_buoc += 1
    if n > gia_tri_max:
        gia_tri_max = n
print("\n--- KẾT QUẢ ---")
print(f"Số xuất phát: {n_ban_dau}")
print(f"Số bước thực hiện: {so_buoc}")
print(f"Giá trị lớn nhất: {gia_tri_max}")
```

```python
import random
random.seed(2026)
danh_sach_N = [100, 10000, 1000000]
PI_THUC_TE = 3.1415926535  

print("=== ƯỚC LƯỢNG PI BẰNG MONTE CARLO ===")

for N in danh_sach_N:
    dem_trong_tron = 0
    for i in range(N):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if (x * x + y * y) <= 1:
            dem_trong_tron = dem_trong_tron + 1
    pi_uoc_luong = 4 * dem_trong_tron / N
    sai_so = abs(pi_uoc_luong - PI_THUC_TE)
    
    print("N =", N, "| Pi ước lượng:", pi_uoc_luong, "| Sai số:", sai_so)
    nguong_cho_phep = 6 / (N ** 0.5)
    assert sai_so < nguong_cho_phep, "Sai số vượt ngưỡng cho phép!"

print("\nNhận xét: Ta thấy gấp 100 lần số điểm (từ 10^2 lên 10^4, hoặc 10^4 lên 10^6) thì sai số giảm đi khoảng 10 lần (đúng bằng căn bậc hai của 100).")
```

```python
def f(x):
    return x**3 - x - 2
a = 1.0
b = 2.0
sai_so_cho_phep = 1e-10
so_buoc = 0

print("--- PHƯƠNG PHÁP CHIA ĐÔI ---")
while (b - a) >= sai_so_cho_phep:
    mid = (a + b) / 2
    so_buoc += 1
    if f(mid) == 0:
        break
    elif f(a) * f(mid) < 0:
        b = mid
    else:
        a = mid

nghiem = (a + b) / 2
import math
so_buoc_ly_thuyet = math.ceil(math.log2(1 / sai_so_cho_phep))

print(f"Nghiệm xấp xỉ của phương trình: {nghiem:.10f}")
print(f"Số bước thực hiện thực tế: {so_buoc}")
print(f"Số bước tính theo lý thuyết: {so_buoc_ly_thuyet}")
print("-> Nhận xét: Số bước thực tế hoàn toàn khớp và nằm trong phần lý thuyết.")
```
